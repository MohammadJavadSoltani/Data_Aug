## save bands from GEE as text, and run this:

```python
import re

# Path to your input file
input_file = 'F:/_____My_Thesies____/_ImpFiles/3. Feature Selection/_Sentinel Data/1. Spec_SAR_Data/2024 Bands Names_Raw.txt'

# Lists to hold cleaned band names
cleaned_bands = []

with open(input_file, "r") as f:
    lines = f.readlines()

for line in lines:
    # Extract the quoted string
    match = re.search(r'"([^"]+)"', line)
    if not match:
        continue

    name = match.group(1)

    if name.startswith("2_"):  # Sentinel-1 (SAR)
        parts = name.split("_")
        if len(parts) == 4:
            band = parts[2]
            date = parts[3]
            cleaned_bands.append(f"{band}_{date}")

    elif name.startswith("1_"):  # Sentinel-2 (Optical)
        # Match band like B2, B3, etc. and the date at the end
        match = re.search(r'_(B\d+A?)_(\d{8})$', name)
        if match:
            band = match.group(1)
            date = match.group(2)
            cleaned_bands.append(f"{band}_{date}")

# Output the result
for band in cleaned_bands:
    print(band)
```
